```python
# imports
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sb

#froms
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler, LabelEncoder, Normalizer
from sklearn.metrics import silhouette_score
```

```python
# globales
df = pd.read_csv("../../data/dataset_training.csv", encoding="ISO-8859-1", low_memory=False)
dfPredicReal = pd.read_csv("../../data/dataset_prediction.csv", encoding="ISO-8859-1", low_memory=False)
print(f'\n DF limpio:')
print(df.info())
```

```python
mask = df['Country'] == "United Kingdom"
df = df[mask]
df = df.set_index("Unnamed: 0")
```

```python
df.head()
```

```python
#config
df_grouped = df.groupby(['Quantity', 'StockCode'])['TotalSales'].sum().reset_index()
df_grouped

df_to_use = df_grouped.groupby(['StockCode']).agg({
  "TotalSales": "sum",
   "Quantity": "sum"
}).reset_index()

df_to_use
```

```python
camposUsar = df_to_use[['Quantity', 'TotalSales']]

# Aplicar K-Means
kmeans = KMeans(n_clusters=4, random_state=42, n_init=10)
df_to_use["Cluster"] = kmeans.fit_predict(camposUsar)

# Agregar las etiquetas al dataframe
df_to_use["Cluster"] = kmeans.labels_

# Evaluar la inercia del modelo
kmeans.inertia_
```

```python
plt.figure(figsize=(8, 6))
plt.scatter(camposUsar["TotalSales"], camposUsar["Quantity"], c=df_to_use["Cluster"], cmap='viridis', alpha=0.7)
plt.xlabel("Total de ventas generado")
plt.ylabel("Numero vendido")
plt.title("Clusters de productos con K-Means")
plt.grid()
plt.show()
```
