```python
# imports
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sb

#froms
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler, LabelEncoder, Normalizer
from sklearn.metrics import silhouette_score
```

```python
# globales
df = pd.read_csv("../../data/dataset_training.csv", encoding="ISO-8859-1", low_memory=False)
dfPredicReal = pd.read_csv("../../data/dataset_prediction.csv", encoding="ISO-8859-1", low_memory=False)
print(f'\n DF limpio:')
print(df.info())
#scaling_factor = 1000 

```

```python
mask = df['Country'] == "United Kingdom"
df = df[mask]
df = df.set_index("Unnamed: 0")
```

```python
df.head()
```

```python
#config 1
df_grouped = df.groupby(['InvoiceNo', 'CustomerID'])['TotalSales'].sum().reset_index()
df_grouped

df_to_use = df_grouped.groupby(['CustomerID']).agg({
  "TotalSales": "sum",
  "InvoiceNo": "count"
}).reset_index()

df_to_use
```

```python
#config 2
df_grouped = df.groupby(['Quantity', 'StockCode'])['TotalSales'].sum().reset_index()
df_grouped

df_to_use = df_grouped.groupby(['StockCode']).agg({
  "TotalSales": "sum",
   "Quantity": "sum"
}).reset_index()

df_to_use
```

```python
df_to_use = df_to_use.sort_values('InvoiceNo')
df_to_use
```

```python
#config 1
camposUsar = df_to_use[['TotalSales', 'InvoiceNo']]

# Aplicar K-Means
kmeans = KMeans(n_clusters=4, random_state=42, n_init=10)
df_to_use["Cluster"] = kmeans.fit_predict(camposUsar)

# Agregar las etiquetas al dataframe
df_to_use["Cluster"] = kmeans.labels_

# Evaluar la inercia del modelo
kmeans.inertia_
```

```python
plt.figure(figsize=(8, 6))
plt.scatter(camposUsar["TotalSales"], camposUsar["InvoiceNo"], c=df_to_use["Cluster"], cmap='viridis', alpha=0.7)
plt.scatter(kmeans.cluster_centers_[:, 0], kmeans.cluster_centers_[:, 1], 
            s=300, c='red', marker='X', label='Centroides')

plt.xlabel("Total de Ventas")
plt.ylabel("Numero de Facturas")
plt.title("Clusters de Clientes con K-Means")
plt.legend()
plt.grid()
plt.show()
```

```python
# config 2
camposUsar = df_to_use[['Quantity', 'TotalSales']]

# Aplicar K-Means
kmeans = KMeans(n_clusters=4, random_state=42, n_init=10)
df_to_use["Cluster"] = kmeans.fit_predict(camposUsar)

# Agregar las etiquetas al dataframe
df_to_use["Cluster"] = kmeans.labels_

# Evaluar la inercia del modelo
kmeans.inertia_
```

```python
plt.figure(figsize=(8, 6))
plt.scatter(camposUsar["TotalSales"], camposUsar["Quantity"], c=df_to_use["Cluster"], cmap='viridis', alpha=0.7)
plt.xlabel("Total de Ventas")
plt.ylabel("Numero vendido")
plt.title("Clusters de Clientes con K-Means")
plt.legend()
plt.grid()
plt.show()
```

```python
# Rango de valores de k que quieres probar
k_values = range(1, 100)  # Puedes ajustarlo según lo necesites
inertia_values = []

# Calcular inercia para cada k
for k in k_values:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(df_to_use)
    inertia_values.append(kmeans.inertia_)
```

```python

best_k = None
best_silhouette = -1
silhouette_scores = []

for k in range(2, 20):  # Evaluamos entre 2 y 20 clusters
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = kmeans.fit_predict(camposUsar)
    silhouette = silhouette_score(camposUsar, labels)
    silhouette_scores.append(silhouette)
    
    if silhouette > best_silhouette:
        best_silhouette = silhouette
        best_k = k

# Graficar el coeficiente de silueta
plt.figure(figsize=(8, 5))
plt.plot(range(2, 20), silhouette_scores, marker='o', linestyle='-', color='g')
plt.xlabel('Número de Clusters (k)')
plt.ylabel('Coeficiente de Silueta')
plt.title('Coeficiente de Silueta para diferentes valores de k')
plt.grid()
plt.show()

print(f"El mejor número de clusters según la silueta es k={best_k} con un valor de {best_silhouette:.4f}")
```
