- Clustering para Reino Unido

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.cluster import DBSCAN

df = pd.read_csv("../../data/dataset_uk_for_clustering.csv")

features_to_use = df[['TotalSales', 'InvoiceNo', 'MediaTicket']]

dbscan = DBSCAN(eps=550, min_samples=5)

df["Cluster"] = dbscan.fit_predict(features_to_use)

plt.figure(figsize=(6, 4))
sns.scatterplot(
  x='InvoiceNo', 
  y='TotalSales', 
  hue='Cluster',
  size='MediaTicket',
  sizes=(20, 200),
  alpha=0.6,
  palette='tab10',
  data=df
)
plt.xlabel("Number of Invoices")
plt.ylabel("Total Sales")
plt.title("Customer Clusters DBSCAN")
plt.legend(loc='lower right')
plt.grid()
plt.show()
```

- Clustering para el resto

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.cluster import DBSCAN

df = pd.read_csv("../../data/dataset_rest_for_clustering.csv")

features_to_use = df[['TotalSales', 'InvoiceNo', 'MediaTicket']]

dbscan = DBSCAN(eps=500, min_samples=5)

df["Cluster"] = dbscan.fit_predict(features_to_use)

plt.figure(figsize=(6, 4))
sns.scatterplot(
  x='InvoiceNo', 
  y='TotalSales', 
  hue='Cluster',
  size='MediaTicket',
  sizes=(20, 200),
  alpha=0.6,
  palette='tab10',
  data=df
)
plt.xlabel("Number of Invoices")
plt.ylabel("Total Sales")
plt.title("Customer Clusters DBSCAN")
plt.legend(loc='lower right')
plt.grid()
plt.show()
```
